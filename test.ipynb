```python
## task 8
import asyncio
import threading

shared_value = 0
lock = threading.Lock()

def cpu_work(name: str):
    """CPU work that runs INSIDE A THREAD.
    Each thread adds values to shared_value, but only with the lock.
    """
    global shared_value
    for i in range(1000):
        with lock:
            shared_value += 1      

    print(f"[{name}] finished CPU work")

async def async_task(name: str):
    """Async task that REQUESTS CPU work and waits for it in a thread."""
    print(f"[{name}] requesting CPU work in a thread...")

    await asyncio.to_thread(cpu_work, name)

async def main():
    print("Start: Async + Thread Sync Hybrid")
    tasks = [
        async_task("Task A"),
        async_task("Task B"),
        async_task("Task C"),
    ]

    await asyncio.gather(*tasks)

    print("All done!")
    print(f"Final shared_value = {shared_value}")

await main()
```

```python
## task 7
import threading
import time

shared_counter = 0

lock = threading.Lock()

def increment_without_lock():
    global shared_counter
    for _ in range(100000):
        
        temp = shared_counter
        time.sleep(0.000001)  
        shared_counter = temp + 1

def increment_with_lock():
    global shared_counter
    for _ in range(100000):
        with lock:
            temp = shared_counter
            time.sleep(0.000001)
            shared_counter = temp + 1

shared_counter = 0
print("Expected counter value: 200000")

t1 = threading.Thread(target=increment_without_lock)
t2 = threading.Thread(target=increment_without_lock)

t1.start()
t2.start()

t1.join()
t2.join()

print("Result WITHOUT lock:", shared_counter)  

shared_counter = 0

t1 = threading.Thread(target=increment_with_lock)
t2 = threading.Thread(target=increment_with_lock)

t1.start()
t2.start()

t1.join()
t2.join()

print("Result WITH lock   :", shared_counter)  
```

```python
## task 6
import asyncio
import time

def heavy_cpu_function():
    print("Starting heavy work...")
    time.sleep(3)  
    print("Finished heavy work!")
    return 100

async def print_message():
    print("Hello from async task!")
    await asyncio.sleep(1)
    print("Async task finished!")

async def main():
    print("Start")

   
    results = await asyncio.gather(
        print_message(),
        asyncio.to_thread(heavy_cpu_function)
    )

    print("Results:", results)
    print("End")


await main()
```

```python
## Task 5
import asyncio

async def worker(name: str, iterations: int):
    """Simulates a task that yields control back to the event loop on each iteration[cite: 31]."""
    for i in range(1, iterations + 1):
        print(f"[{name}] Processing step {i}/{iterations}")
        
       
        await asyncio.sleep(0)

async def main():
    print("--- Starting Cooperative Multitasking Demo ---\n")
    
    await asyncio.gather(
        worker("Task A", 5),
        worker("Task B", 5)
    )
    
    print("\n--- Completed All Tasks ---")
  
  
await main()
```

```python
## Task 4
import asyncio

def heavy_cpu_function():
    print("Starting heavy CPU task...")
    total = 0
    for i in range(50_000_000):
        total += i
    print("Finished heavy CPU task!")
    return total

async def async_task():
    for i in range(3):
        print(f"Async task step {i + 1}")
        await asyncio.sleep(1)

async def main():
    print("Start main")
    
    task = asyncio.create_task(async_task())
    await asyncio.sleep(0.1)  
    heavy_cpu_function()
    
    await task
    print("End main")

await main()
```
